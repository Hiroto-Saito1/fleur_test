```python
import numpy as np
import matplotlib.pylab as plt
import pandas as pd
from io import StringIO
```

```python
# ファイルを読み込みながらDをEに置き換える
with open("DOS.1", "r") as file:
    data1 = file.read().replace('D', 'E')

with open("DOS.2", "r") as file:
    data2 = file.read().replace('D', 'E')

# 置き換えたデータを再度読み込み
dos1 = pd.read_table(StringIO(data1), delimiter="\\s+", header=None, dtype=float)
dos2 = pd.read_table(StringIO(data2), delimiter="\\s+", header=None, dtype=float)
```

```python
plt.figure()
# plt.plot(dos1.iloc[:,1], dos1.iloc[:,0], label="total")
# plt.plot(dos1.iloc[:,2] + dos1.iloc[:,5], dos1.iloc[:,0],)

plt.plot(dos1.iloc[:,2], dos1.iloc[:,0], color="red", label="down")
plt.plot(-dos1.iloc[:,5], dos1.iloc[:,0], color="blue", label="up")

#plt.plot(-dos1.iloc[:,6], dos1.iloc[:,0], label="3s up")
#plt.plot(-dos1.iloc[:,7], dos1.iloc[:,0], label="3p up")
plt.plot(-dos1.iloc[:,8], dos1.iloc[:,0], label="3d up")
#plt.plot(-dos1.iloc[:,9], dos1.iloc[:,0], label="3f up")

#plt.plot(dos2.iloc[:,6], dos2.iloc[:,0], label="3s down")
#plt.plot(dos2.iloc[:,7], dos2.iloc[:,0], label="3p down")
plt.plot(dos2.iloc[:,8], dos2.iloc[:,0], label="3d down")
#plt.plot(dos2.iloc[:,9], dos2.iloc[:,0], label="3f down")


plt.ylim(-9, 5)
plt.xlabel("DOS [states/eV]")
plt.ylabel("$E - E_F$ [eV]")
plt.grid()
plt.legend()
plt.savefig("dos.pdf")
plt.show()
```

```python
up = pd.read_table("Local.1", delimiter="\\s+")
down = pd.read_table("Local.2", delimiter="\\s+")

plt.rcParams["font.size"] = 16
plt.rcParams["figure.subplot.left"] = 0.20
plt.rcParams["figure.subplot.bottom"] = 0.20
plt.figure()

# Total DOS
(up_total,) = plt.plot(up["Total"], up["#energy"], color="red", label="Total (up)")
(down_total,) = plt.plot(
    -down["Total"], down["#energy"], color="blue", label="Total (down)"
)


# Fe 2p
(up_fe2p,) = plt.plot(
    4*up["MT:1p"], up["#energy"], color="orange", label="Fe 2p"
)

(down_fe2p,) = plt.plot(
    -4*down["MT:1p"], down["#energy"], color="green", label="Fe 2p"
)


# O 2p
(up_o2p,) = plt.plot(8 * up["MT:3p"], up["#energy"], color="brown", label="O 2p")
(down_o2p,) = plt.plot(
    -8 * down["MT:3p"], down["#energy"], color="purple", label="O 2p"
)


# Upの凡例
up_legend = plt.legend(
    handles=[
        up_total,
        up_fe2p,
        up_o2p,
    ],
    loc="center right",
    fontsize=10,
)


# Downの凡例
down_legend = plt.legend(
    handles=[
        down_total,
        down_fe2p,
        down_o2p,
    ],
    loc="center left",
    fontsize=10,
)

# Upの凡例をプロットに追加
plt.gca().add_artist(up_legend)

plt.xlabel("DOS [states/eV]")
plt.ylabel("$E - E_F$ [eV]")
plt.ylim(-70, 16)
plt.grid()
plt.savefig("dos.pdf")
plt.show()
```
